```python
import pandas as pd
```

```python
df=pd.read_csv("C:\\Users\\Deepika\\Desktop\\ML_MODEL_TRAINING\\Project_8_Dataset.csv")
```

```python
print(df.isnull().sum())
```

```python

missing_percent = (df.isnull().sum() / len(df)) * 100

missing_df = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Percentage': missing_percent
})

missing_df = missing_df[missing_df['Missing Values'] > 0]

missing_df.sort_values(
    by='Percentage',
    ascending=False
)
```

```python


weather_columns = [
    "temperature_c",
    "humidity_pct",
    "wind_kmph",
    "pressure_hpa",
    "cloud_cover_pct",
    "rain_probability_pct"
]

print("WEATHER DATASET AVERAGES")

print(df[weather_columns].mean())

print("\nMINIMUM VALUES")
print(df[weather_columns].min())

print("\nMAXIMUM VALUES")
print(df[weather_columns].max())

print("\nSTANDARD DEVIATION")
print(df[weather_columns].std())
```
